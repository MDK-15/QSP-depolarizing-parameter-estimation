The goal of this notebook is to see how the achieved sensitivity (kappa, L_E) varies with the circuit depth (d). As shown below, the optimal value of d depends on the value of p. Concretely, lower values of p can benefit from higher circuit depth, while higher of p should be kept at lower circuit depth to achieve the best sensitivity.

```python
from QSP_PE import QSP_PE
from single_iteration import QSP_single_iteration
import matplotlib.pyplot as plt
import numpy as np
from estimateP import estimate_p_bisect
from noise_signal_design import optimize_noise_sensitivity
from single_iteration import *
```

```python
zero = np.array([1, 0], dtype=complex)
rho = np.outer(zero, zero.conj().T)

p_real = 0.001
p_prev = 0.0015
dk = [2, 4, 8, 16, 32, 64]
C = 5
q = 2
rk_prev = 0.001
mk = 8000
```

```python
def QSP_single_iteration_sensitivity_depth(mk, C, dk, rk_prev, q, p_real, p_prev, rho):
    results_kappa = []
    results_L_E = []
    for d in dk:
        theta, Phi, L_E, kappa = optimize_noise_sensitivity(d=d, p_lo=p_prev-rk_prev, p_hi=p_prev+rk_prev)
        results_kappa.append(kappa)
        results_L_E.append(L_E)
    return results_kappa, results_L_E

kappa, L_E = QSP_single_iteration_sensitivity_depth(mk=mk, C=C, dk=dk, rk_prev=rk_prev, q=q, p_real=p_real, p_prev=p_prev, rho=rho)
```

```python
fig, axs = plt.subplots(1, 2, figsize=(16, 4.5))
axs[0].plot(dk, kappa, "o-", lw=2, markersize=5)
axs[0].set_xticks(dk)
axs[0].set_xlabel('Circuit depth')
axs[0].set_title('kappa')
axs[0].grid(True, alpha=0.3)

axs[1].plot(dk, L_E, "o-", lw=2, markersize=5)
axs[1].set_xticks(dk)
axs[1].set_xlabel('Circuit depth')
axs[1].set_title('L_E')
axs[1].grid(True, alpha=0.3)

plt.show()
```
